### Вопрос 1
**Что такое переменная в Python? Присваивание, корректные и некорректные имена.**

Переменная в Python — это именованная ссылка на объект в оперативной памяти, а не фиксированная ячейка. Оператор `=` связывает имя объекта с самим объектом.

- **Корректные имена:** `user_age`, `_temp`, `total_price2` (состоят из букв, цифр и подчёркиваний, не начинаются с цифры).
- **Некорректное имя:** `2nd_value` — имя не может начинаться с цифры, так как синтаксический анализатор воспринимает первый символ как числовой литерал.

### Вопрос 2
**Основные простейшие типы данных, литералы, type() и динамическая типизация.**

1. `int` — целые числа (литерал: `42`), `type(42)` -> `<class 'int'>`
2. `float` — числа с плавающей точкой (литерал: `3.14`), `type(3.14)` -> `<class 'float'>`
3. `str` — строки (литерал: `'тест'`), `type('тест')` -> `<class 'str'>`
4. `bool` — булевы значения (литерал: `True`), `type(True)` -> `<class 'bool'>`

**Динамическая типизация:** тип привязан к объекту, а не к переменной. Переменная может в любой момент сослаться на объект другого типа без явного объявления.

```python
# Демонстрация типов
print(type(42), type(3.14), type("тест"), type(True))
```

`print(*objects, sep=' ', end='\n')` преобразует переданные объекты в строки и выводит их в стандартный поток вывода.

`type(obj)` возвращает класс, экземпляром которого является переданный объект. Результат `<class 'X'>` показывает встроенный тип Python.

### Вопрос 5
**Арифметические операции в Python и типы результатов.**

- Сложение `+`, вычитание `-`, умножение `*`: дают `int`, если оба операнда `int`; дают `float`, если хотя бы один `float`.
- Обычное деление `/`: **всегда** возвращает `float` (даже `4 / 2` -> `2.0`).
- Целочисленное деление `//` и остаток `%`: сохраняют целочисленный тип для двух `int`, возвращают `float`, если есть `float`.
- Возведение в степень `**`: дает `float` при отрицательной степени или операнде `float`.

```python
print(10 + 2, type(10 + 2))        # int
print(10 / 2, type(10 / 2))        # float (всегда)
print(10 // 3, type(10 // 3))      # int
print(10 % 3, type(10 % 3))        # int
print(2 ** 3, type(2 ** 3))        # int
print(2 ** -1, type(2 ** -1))      # float
```

### Вопрос 6
**Разница между `/` и `//`. Положительные и отрицательные числа.**

- `/` — точное деление (float).
- `//` — деление с округлением вниз до ближайшего меньшего целого (floor division).
  - Для положительных: `7 // 2 == 3` (3.5 округляется к 3).
  - Для отрицательных: `-7 // 2 == -4` (-3.5 округляется к меньшей бесконечности, то есть к -4, а не к -3).

```python
print("Положительные:", 7 / 2, "| // :", 7 // 2)
print("Отрицательные:", -7 / 2, "| // :", -7 // 2)
print("-10 // 3:", -10 // 3)
```

### Вопрос 7
**Преобразование типов: int(), float(), str() и возможные ошибки.**

- `int("123")` -> `123`, `float("3.14")` -> `3.14`, `str(100)` -> `"100"`.
- **Ошибки:** вызов `int("abc")` или `int("3.14")` вызывает `ValueError`, так как строка не является строковым представлением целого числа в десятичной системе счисления. Для `"3.14"` нужно сначала вызвать `float()`, затем `int()`.

```python
a = int("45")
b = float("12.8")
c = str(500)
print(a, b, c)

try:
    int("hello")
except ValueError as e:
    print("Ошибка при int('hello'):", e)
```

### Вопрос 8
**«Вычисление в Python — это выражение, возвращающее значение». Примеры.**

Любое выражение (expression) при вычислении редуцируется в конкретный объект-значение, который можно сразу присвоить переменной или передать аргументом в функцию.

### Вопрос 9
**Операторы сравнения и логические операторы. Приоритеты и составные выражения.**

Приоритет: Арифметика > Операторы сравнения (`==`, `<`, `>` и др.) > `not` > `and` > `or`.

```python
# Выражение 1: 5 > 2 (True) and not 3 == 4 (not False -> True) -> True
exp1 = 5 > 2 and not 3 == 4

# Выражение 2: 10 < 5 (False) or 2 + 2 == 4 (True) -> True
exp2 = 10 < 5 or 2 + 2 == 4

print("exp1:", exp1)
print("exp2:", exp2)
```

### Вопрос 10
**Операции над операндами разных типов: неявное vs явное приведение.**

- **Неявное:** `int + float -> float` (например, `2 + 3.0 -> 5.0`), `bool + int -> int` (`True + 1 -> 2`).
- **Требуется явное:** сложение строк и чисел (`"code" + 5`) вызывает `TypeError`. Python строго типизирован и не приводит строки к числам автоматически.

```python
print(3 + 2.5)       # 5.5 (неявное к float)
print(True + 10)     # 11 (неявное bool -> int)

# print("Count: " + 5) # TypeError
print("Count: " + str(5)) # Явное приведение
```

### Вопрос 11
**Способы вывода через print(): сравнение.**

1. **Через запятую:** быстро для отладки, разделяет пробелами.
2. **Конкатенация (`+`):** требует ручного `str()`, неудобна при многих переменных.
3. **Метод `.format()`:** удобен при отделении шаблона от данных.
4. **f-строки:** стандарт де-факто: лаконичны, поддерживают выражения и форматирование чисел прямо в скобках.

```python
x, y = "Item", 50
print(x, y)                          # Запятая
print(x + ": " + str(y))             # Конкатенация
print("{}: {}".format(x, y))          # .format()
print(f"{x}: {y}")                   # f-строка
```

### Вопрос 12
**Порядок выполнения операций в Python и скобки.**

Приоритет: `**` > унарные `+ -` > `* / // %` > `+ -` бинарные. Скобки `()` имеют наивысший приоритет и принудительно задают последовательность действий.

```python
# 2 + 3 * 4 ** 2 = 2 + 3 * 16 = 2 + 48 = 50
res1 = 2 + 3 * 4 ** 2

# ((2 + 3) * 4) ** 2 = (5 * 4) ** 2 = 20 ** 2 = 400
res2 = ((2 + 3) * 4) ** 2

print("Без скобок:", res1)
print("Со скобками:", res2)
```

### Вопрос 14
**Изменяемые и неизменяемые типы данных. id().**

Числа (`int`, `float`) и строки (`str`) неизменяемы (immutable). Любая операция над ними не модифицирует исходный объект в памяти, а порождает новый с другим адресом (`id`).

```python
s = "hello"
old_id = id(s)
s += " world"
new_id = id(s)

print("id до:", old_id)
print("id после:", new_id)
print("Объект изменился?", old_id != new_id)
```

### Вопрос 15
**Фрагмент кода (5–7 строк) с вычислениями, print() и type().**

1. `base = 15` — целочисленный литерал `int`.
2. `rate = 1.5` — число с плавающей точкой `float`.
3. `total = base * rate` — результат операции умножения `int` на `float` даёт `float` (22.5).
4. `tag = f"ID_{int(total)}"` — явное приведение `total` к `int` (22) и интерполяция в строку `str`.
5. `is_valid = total > 20` — операция сравнения возвращает булево значение `bool` (`True`).
6. Вывод значений и их типов через `print` и `type`.

```python
base = 15
rate = 1.5
total = base * rate
tag = f"ID_{int(total)}"
is_valid = total > 20

print(f"total: {total} ({type(total).__name__})")
print(f"tag: {tag} ({type(tag).__name__})")
print(f"is_valid: {is_valid} ({type(is_valid).__name__})")
```
