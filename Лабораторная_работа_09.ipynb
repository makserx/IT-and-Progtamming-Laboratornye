## 1. Индивидуальное задание (Вариант 6)

**Задание:** Класс «Шар». Реализовать ввод и вывод полей данных, вычисление объема, диаметра и площади поверхности, а также вывод информации об объекте.

### Основные формулы:
- Диаметр: $d = 2r$
- Площадь поверхности: $S = 4\pi r^2$
- Объём: $V = \frac{4}{3}\pi r^3$

```python
import math

class Sphere:
    def __init__(self, radius: float = 1.0):
        self.__radius = 0.0
        self.set_radius(radius)

    # Метод-мутатор (сеттер) для задания радиуса с проверкой
    def set_radius(self, radius: float) -> None:
        if radius <= 0:
            raise ValueError("Радиус шара должен быть положительным числом.")
        self.__radius = float(radius)

    # Метод-получатель (геттер) для получения радиуса
    def get_radius(self) -> float:
        return self.__radius

    # Ввод полей данных с клавиатуры
    def input_data(self) -> None:
        r = float(input("Введите радиус шара: "))
        self.set_radius(r)

    # Вывод полей данных
    def print_radius(self) -> None:
        print(f"Радиус шара: {self.__radius}")

    # Вычисление диаметра
    def get_diameter(self) -> float:
        return 2 * self.__radius

    # Вычисление площади поверхности
    def get_surface_area(self) -> float:
        return 4 * math.pi * (self.__radius ** 2)

    # Вычисление объема
    def get_volume(self) -> float:
        return (4 / 3) * math.pi * (self.__radius ** 3)

    # Строковое представление объекта
    def __str__(self) -> str:
        return (
            f"Шар (r = {self.__radius:.2f}):\n"
            f"  Диаметр: {self.get_diameter():.2f}\n"
            f"  Площадь поверхности: {self.get_surface_area():.2f}\n"
            f"  Объем: {self.get_volume():.2f}"
        )

    # Вывод полной информации об объекте
    def info(self) -> None:
        print(self)
```

```python
# Демонстрация работы класса Sphere
sphere1 = Sphere(5)
sphere1.print_radius()
sphere1.info()

# Изменение радиуса через сеттер
sphere1.set_radius(3)
print("\nПосле изменения радиуса:")
print(f"Диаметр: {sphere1.get_diameter():.2f}")
print(f"Площадь поверхности: {sphere1.get_surface_area():.2f}")
print(f"Объем: {sphere1.get_volume():.2f}")

# Для интерактивного ввода с клавиатуры раскомментируйте строки ниже:
# sphere2 = Sphere()
# sphere2.input_data()
# sphere2.info()
```

## 2. Ответы на контрольные вопросы

### 1. Дайте определение понятию класс.
**Класс** — это пользовательский тип данных (шаблон или чертеж), который объединяет данные (атрибуты/поля) и функции для работы с этими данными (методы) в единую структуру. На основе класса создаются конкретные экземпляры (объекты).

---

### 2. Продемонстрируйте использование метода `__init__`. Поясните назначение данного метода.
Метод `__init__` — это инициализатор (конструктор) класса. Он автоматически вызывается сразу после создания нового объекта и нужен для задания начальных значений его атрибутов.

```python
class Point:
    def __init__(self, x, y):
        self.x = x  # инициализация атрибута x
        self.y = y  # инициализация атрибута y

p = Point(3, 7)  # __init__ вызывается автоматически
print(f"Точка: ({p.x}, {p.y})")
```

### 3. Поясните назначение статических методов и их отличия от методов класса.
- **Статический метод (`@staticmethod`)** — это обычная функция, помещенная в пространство имен класса для логической группировки. Она не принимает ни `self` (ссылку на экземпляр), ни `cls` (ссылку на класс) и не имеет доступа к состоянию объекта или класса.
- **Метод класса (`@classmethod`)** — принимает первым аргументом сам класс (`cls`), поэтому может обращаться к атрибутам класса, изменять их и создавать новые экземпляры класса (например, альтернативные конструкторы).

```python
class Circle:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    @staticmethod
    def calc_diameter(r):
        return 2 * r  # не зависит от cls и self

    @classmethod
    def from_diameter(cls, d):
        return cls(d / 2)  # использует cls для создания объекта

print(Circle.calc_diameter(5))
c = Circle.from_diameter(8)
print(c.radius)
```

### 4. Для чего используется параметр `self` в каждом методе класса?
Параметр `self` хранит ссылку на конкретный экземпляр класса, для которого вызван метод. Через `self` метод получает доступ к атрибутам и другим методам именно этого объекта.

---
### 5. Сколько методов `__init__` может быть объявлено внутри класса?
В Python внутри одного класса может быть только **один** активный метод `__init__`. Перегрузка методов по сигнатуре не поддерживается: если объявить несколько `__init__`, каждое следующее определение перезапишет предыдущее. Для разных вариантов инициализации используют значения параметров по умолчанию или `@classmethod`.

---
### 6. Каким образом в классе Python атрибут скрывается от программного кода, находящегося за пределами класса?
Для сокрытия (инкапсуляции) атрибута перед его именем ставится **двойное подчеркивание** (`__attribute`). В этом случае срабатывает механизм искажения имен (*name mangling*): внутри класса атрибут доступен как `self.__attribute`, а снаружи переименовывается в `_ClassName__attribute`, что защищает его от случайного прямого доступа.

---
### 7. Какова задача метода `__str__()`?
Метод `__str__()` возвращает читаемое строковое представление объекта, понятное пользователю.

---
### 8. Каким образом происходит вызов метода `__str__()`?
Метод `__str__()` вызывается автоматически при передаче объекта в встроенные функции `print()`, `str()` или при форматировании строк (например, в f-строках `f"{obj}"`).

---
### 9. Что такое атрибут экземпляра?
**Атрибут экземпляра** — это переменная, принадлежащая конкретному объекту (экземпляру) класса. Обычно создается внутри `__init__` через `self.имя_атрибута` и хранит индивидуальное состояние каждого объекта (в отличие от атрибутов класса, общих для всех экземпляров).

---
### 10. Что такое метод-получатель? Что такое метод-мутатор?
- **Метод-получатель (геттер, getter)** — метод, который возвращает значение закрытого атрибута, не изменяя состояние объекта (например, `get_radius()`).
- **Метод-мутатор (сеттер, setter)** — метод, который устанавливает или изменяет значение закрытого атрибута, обычно с проверкой корректности данных (например, `set_radius(value)`).
