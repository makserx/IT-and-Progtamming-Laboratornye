**Условие:** Пользователь вводит число. Программа определяет, положительное оно, отрицательное или равно нулю, и выводит результат. Используйте условный оператор.

### Уровень 2 – Средний (Вариант 6)
**Условие:** Пользователь вводит число $n$. Программа определяет, является ли оно совершенным (равно сумме всех своих делителей, кроме самого себя). Используйте цикл `for` и условный оператор. Пример: $6 = 1 + 2 + 3$.

```python
# n = int(input("Введите натуральное число n: "))
n = 28  # Тестовое совершенное число: 1 + 2 + 4 + 7 + 14 = 28

divisors_sum = 0
for i in range(1, n):
    if n % i == 0:
        divisors_sum += i

if divisors_sum == n and n > 0:
    print(f"Число {n} является совершенным (сумма собственных делителей = {divisors_sum})")
else:
    print(f"Число {n} не является совершенным (сумма делителей = {divisors_sum})")
```

### Уровень 3 – Продвинутый (Вариант 6)
**Условие:** Пользователь вводит число $n$. Программа вычисляет золотое сечение $\phi = (1 + \sqrt{5})/2$ двумя способами: через `math.sqrt` и через предел последовательности Фибоначчи $F_{n+1}/F_n$. Выведите оба результата с 10 знаками и разность между ними.

```python
import math

# n = int(input("Введите n (например, 40): "))
n = 40  # Оптимальное значение для высокой сходимости ряда

# Способ 1: Аналитическая формула через math.sqrt
phi_formula = (1 + math.sqrt(5)) / 2

# Способ 2: Предел отношения соседних чисел Фибоначчи F_{n+1} / F_n
f_prev, f_curr = 1, 1
for _ in range(n - 1):
    f_prev, f_curr = f_curr, f_prev + f_curr

phi_fib = f_curr / f_prev
diff = abs(phi_formula - phi_fib)

print(f"phi (аналитический через math.sqrt): {phi_formula:.10f}")
print(f"phi (предел Фибоначчи F_{{n+1}}/F_n): {phi_fib:.10f}")
print(f"Разность между результатами:          {diff:.10e}")
```

### Вопрос 1
**В чём принципиальное отличие циклов for и while в Python? Приведите по одному примеру задачи, где каждый из них будет наиболее уместен, и кратко обоснуйте выбор.**

- `for` применяется тогда, когда количество итераций детерминировано или известно заранее, либо требуется перебрать элементы готовой последовательности (список, кортеж, строка, генератор `range`).
- `while` применяется, когда количество итераций неизвестно заранее и цикл должен повторяться до выполнения некоторого динамического логического условия.

**Примеры задач:**
- Для `for`: расчёт суммы чисел массива или вывод таблицы умножения (`range(1, 11)`). Выбор обусловлен гарантированным конечным числом шагов.
- Для `while`: приближённое вычисление корня методом Герона или ряда Тейлора до достижения погрешности $\varepsilon$, либо цикл повторного ввода логина до успешной аутентификации. Число шагов зависит от данных во время работы программы.

```python
# Пример for: 5 фиксированных шагов
for step in range(1, 6):
    pass

# Пример while: завершение по условию сходимости
val, eps = 1.0, 0.05
while val > eps:
    val /= 2
print("while завершён, val:", val)
```

### Вопрос 2
**Как работает блок else внутри цикла (for/while)? Опишите, при каких условиях он выполняется, а при каких – нет. Приведите два коротких примера кода: в одном else срабатывает, в другом – нет (из за break).**

Блок `else` в цикле выполняется только в том случае, если цикл завершился штатным путём (последовательность в `for` исчерпана или условие в `while` стало ложным). Если выход из цикла произошёл досрочно по оператору `break`, ветка `else` игнорируется.

```python
# Пример 1: else срабатывает (break не был вызван)
for x in [1, 3, 5]:
    if x % 2 == 0:
        break
else:
    print("Чётных элементов нет (блок else сработал)")

# Пример 2: else НЕ срабатывает (выход по break)
for x in [1, 4, 5]:
    if x % 2 == 0:
        print(f"Найдено чётное {x}, выходим досрочно")
        break
else:
    print("Этот блок не выполнится")
```

### Вопрос 3 (п. 4 в метод. указаниях)
**Объясните разницу между операторами break и continue. Напишите фрагмент кода с циклом while, где используются оба оператора, и прокомментируйте, что произойдёт на каждой итерации.**

- `break` прерывает выполнение цикла целиком и выводит поток управления за его пределы.
- `continue` прекращает текущую итерацию цикла и переходит к проверке условия следующего шага.

```python
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue  # Чётные i (2, 4, 6...) пропускают вывод и сразу идут на i += 1
    if i > 6:
        break     # При i = 7 цикл завершается полностью
    print(f"Обработано нечётное число: {i}")
```

### Вопрос 5 (п. 6 в метод. указаниях)
**Как устроена условная конструкция if/elif/else? Напишите пример, где проверяется принадлежность числа диапазонам, и объясните, почему порядок условий важен.**

Конструкция `if/elif/else` выполняет последовательную проверку условий сверху вниз. Как только одно из условий возвращает `True`, исполняется связанный блок кода, а все последующие ветки игнорируются.

**Почему порядок важен:** если поставить более общее условие выше частного, оно перехватит выполнение, и частный блок станет недостижимым.

```python
val = 7
if val < 0:
    print("меньше 0")
elif val <= 10:
    print("от 0 до 10")
else:
    print("больше 10")
```

### Вопрос 6 (п. 7 в метод. указаниях)
**Что такое логические операторы and, or, not и как они влияют на работу условий? Составьте единое выражение по заданию.**

- `and` (логическое И) — истинно только тогда, когда оба операнда истинны.
- `or` (логическое ИЛИ) — истинно, если хотя бы один операнд истинен.
- `not` (логическое НЕ) — инвертирует логическое значение.

**Выражение:** число чётное и в $[10, 20]$, либо нечётное и больше 100:

```python
condition = lambda n: ((n % 2 == 0) and (10 <= n <= 20)) or ((n % 2 != 0) and (n > 100))

print("14:", condition(14))    # True (чётное от 10 до 20)
print("105:", condition(105))  # True (нечётное > 100)
print("8:", condition(8))      # False
print("102:", condition(102))  # False (чётное > 100)
```

### Вопрос 7 (п. 8 в метод. указаниях)
**Как в Python происходит вычисление арифметических выражений? Объясните приоритет операций и покажите, как скобки меняют порядок вычислений.**

Приоритет: возведение в степень `**` (ассоциативность справа налево) $\to$ унарные `+`, `-` $\to$ мультипликативные `*`, `/`, `//`, `%` $\to$ аддитивные `+`, `-`. Круглые скобки `()` обладают наивысшим приоритетом.

```python
# 1. 2 + 3 * 2 ** 3 = 2 + 3 * 8 = 2 + 24 = 26
res1 = 2 + 3 * 2 ** 3
# 2. ((2 + 3) * 2) ** 3 = (5 * 2) ** 3 = 10 ** 3 = 1000
res2 = ((2 + 3) * 2) ** 3
# 3. 20 // 6 % 2 = 3 % 2 = 1
res3 = 20 // 6 % 2

print("res1:", res1, "| res2:", res2, "| res3:", res3)
```

### Вопрос 8 (п. 9 в метод. указаниях)
**Чем отличаются операторы деления / и //? Напишите код и поясните типы.**

- `/` — деление с плавающей точкой, результат **всегда** имеет тип `float`.
- `//` — целочисленное деление с округлением вниз (floor). Если оба аргумента `int`, результат будет `int`. На отрицательных числах округление идёт в сторону меньшего целого (например, `-7 // 2 = -4`).

```python
cases = [(9, 3), (7, 2), (-7, 2), (7.0, 2)]
for a, b in cases:
    f_div = a / b
    i_div = a // b
    print(f"{a} / {b} = {f_div:<5} ({type(f_div).__name__}) | {a} // {b} = {i_div:<5} ({type(i_div).__name__})")
```

### Вопрос 9 (п. 10 в метод. указаниях)
**Как работают встроенные функции min, max, sum, abs, round?**

- `min(iterable | *args)`: возвращает минимальный элемент.
- `max(iterable | *args)`: возвращает максимальный элемент.
- `sum(iterable, start=0)`: возвращает сумму элементов числовой последовательности.
- `abs(x)`: возвращает абсолютную величину числа (тип совпадает с типом аргумента).
- `round(x, n)`: выполняет округление (банковское — к ближайшему чётному). Без аргумента $n$ возвращает `int`, с $n$ — `float`.

```python
nums = [4, -8, 15, 23, -42]
print("min:", min(nums))
print("max:", max(nums))
print("sum:", sum(nums))
print("abs:", abs(-17.5))
print("round:", round(2.675, 2), round(5.5))
```

### Вопрос 12 (п. 13 в метод. указаниях)
**Как вычислить квадратный корень в Python? Сравнение трёх способов.**

- `math.sqrt(x)`: быстрая реализация на C; при $x < 0$ выбрасывает `ValueError` (строгий контроль вещественной области).
- `x ** 0.5`: компактный синтаксис; при $x < 0$ не падает, а переходит в комплексные числа (`complex`), что иногда маскирует ошибки логики.
- **Приближённые методы (Герон):** алгоритмически наглядны для изучения численных методов, но медленнее и требуют остановки по порогу точности.

```python
import math
x = 49

# 1. math.sqrt
r1 = math.sqrt(x)
# 2. Степень 0.5
r2 = x ** 0.5
# 3. Метод Герона
y = float(x)
while abs(y * y - x) > 1e-10:
    y = (y + x / y) / 2

print(f"math.sqrt: {r1} | **0.5: {r2} | Heron: {y}")
```

### Вопрос 13 (п. 14 в метод. указаниях)
**Как обрабатывать «проблемные» случаи в вычислениях? Безопасный корень.**

Граничные случаи проверяются до выполнения операции с помощью конструкции `if/else`, исключая падение программы.

```python
val_str = "-9"
try:
    number = float(val_str)
    if number < 0:
        print(f"Ошибка: число {number} отрицательное, вещественный корень не определён")
    else:
        print(f"Квадратный корень: {number ** 0.5}")
except ValueError:
    print("Ошибка: введено нечисловое значение")
```

### Вопрос 15 (п. 16 в метод. указаниях)
**Как выбрать между ручным подсчётом в цикле и использованием встроенных функций?**

- Встроенные функции (`sum`, `min`, `max`, `len`) написаны на языке Си, выполняются на порядок быстрее, делают код компактным и защищённым от человеческих ошибок (забытые счётчики, некорректная инициализация аккумулятора).
- Ручной цикл необходим при нестандартной агрегации (одновременный сбор статистики, сложная фильтрация, прерывание по условию).

```python
data = [5, 12, 7, 3, 20]

# Вариант 1: Ручной цикл
acc_sum = 0
acc_min = data[0]
for val in data:
    acc_sum += val
    if val < acc_min:
        acc_min = val

# Вариант 2: Встроенные функции
fast_sum = sum(data)
fast_min = min(data)

print("Ручной подсчёт:    sum =", acc_sum, ", min =", acc_min)
print("Встроенные функции: sum =", fast_sum, ", min =", fast_min)
```
