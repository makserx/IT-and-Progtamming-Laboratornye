## Пример из лабораторной работы

В файле `pr5.txt` содержится последовательность целых чисел от –10 000 до 10 000 включительно. Определите количество пар, в которых оба элемента больше среднего арифметического всех чисел, а их сумма оканчивается на 31. В ответе запишите количество пар и максимальную сумму.

````python
!wget -q -nc https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
````

````python
# Чтение данных примера
data = []
with open('pr5.txt') as f:
    for line in f:
        data.append(int(line))

sr = sum(data) / len(data)

counter = 0
max_sum = -10000
for i in range(len(data) - 1):
    x, y = data[i], data[i + 1]
    if x > sr and y > sr and str(x + y)[-2:] == '31':
        counter += 1
        max_sum = max(max_sum, x + y)

print(f'Пример — Количество пар: {counter}')
print(f'Пример — Максимальная сумма: {max_sum}')
````

## Индивидуальное задание

**Условие:**
В файле `pr5_1.txt` содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от 0 до 10 000 включительно. Определите количество пар, в которых оба элемента меньше, чем среднее арифметическое всех чисел в файле, и хотя бы один из двух элементов оканчивается на 9. В ответе запишите два числа: сначала количество найденных пар, а затем – максимальную сумму элементов таких пар. В данной задаче под парой подразумевается два идущих подряд элемента последовательности.

````python

!wget -q -nc https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_1.txt
````

````python
with open('pr5_1.txt') as f:
    s = [int(x) for x in f]

avg = sum(s) / len(s)
pairs = []

for i in range(len(s) - 1):
    a, b = s[i], s[i + 1]
    if a < avg and b < avg and (a % 10 == 9 or b % 10 == 9):
        pairs.append(a + b)

print(len(pairs), max(pairs))
````

**Описание списковых методов:**
- **а) `index(x)`** — возвращает индекс первого вхождения элемента `x` в списке. Если элемент не найден, вызывает ошибку `ValueError`.
- **б) `insert(i, x)`** — вставляет элемент `x` в позицию с индексом `i`, сдвигая последующие элементы вправо.
- **в) `sort()`** — сортирует элементы текущего списка на месте (in-place) по возрастанию (по умолчанию).
- **г) `reverse()`** — разворачивает элементы текущего списка в обратном порядке на месте (in-place).
