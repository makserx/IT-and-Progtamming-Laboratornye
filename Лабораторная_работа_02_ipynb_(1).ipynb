## Задание

Необходимо вычислить значение выражения:
$$Z=\frac{1}{1\cdot3}-\frac{X}{2\cdot4}+\frac{X^2}{3\cdot5}-\frac{X^3}{4\cdot6}+...$$
Для каждой переменной необходимо вывести ее значение и тип.
$$s =-\frac{Y\cdot X^2}{1\cdot3}+\frac{X\cdot Y^3}{2\cdot4}-\frac{Y\cdot X^4}{3\cdot5}+\frac{X\cdot Y^5}{4\cdot6}-...$$

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X$. Переменная, которая должна быть вычислена на основе независимых: $Z$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

Обратите внимание, что знаки слагаемых чередуются, а степени числителя и множетели знаменателя привязаны к порядковому номеру слагаемого.

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

````python
X = 2
N = 10
Z = 0
for i in range(1, N+1):
  chisl = X**(i-1)
  znam = i * (i+1)
  if i%2==1:
    Z += chisl / znam
  else:
    Z -= chisl/znam
print(Z)
````

Решите пример в соответствии с индивидуальным заданием

````python
# Здесь код индивидуального задания...
s = 0
N = 10
X = 2
Y = 3

for i in range(1, N+1):
    if i % 2 == 0:
        s += (X * Y ** (i+1)) / ((i+1) * (i+2))
    else:
        s -= (Y * X ** (i+1)) / ((i+1) * (i+2))

print(s)
````

13. Что такое накопитель (аккумуляторная переменная)?
Это переменная, которая используется для накопления результата (суммы, произведения, количества и т.п.) в цикле. Её значение обновляется на каждой итерации на основе предыдущего значения.

14. Следует ли инициализировать накопитель конкретным значением?
Да, следует. Накопитель нужно инициализировать до начала цикла (например, total = 0 для суммы), иначе при первом обращении к нему возникнет ошибка NameError, либо результат будет неверным.

20. Что такое бесконечный цикл? Пример кода.
Бесконечный цикл — это цикл, который выполняется непрерывно, пока не будет прерван извне (например, оператором break или аварийным завершением программы).

Пример:
```python

while True:
    print('Этот цикл бесконечный')
```
