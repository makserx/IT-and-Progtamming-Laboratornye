## Задание

Текстовый файл `pr6_1.txt` состоит не более чем из $10^6$ символов и содержит только заглавные буквы латинского алфавита и десятичные цифры. Найдите количество букв `D`, перед которыми находятся две цифры.

## Решение

````python
# Команда для скачивания файла при работе в Google Colab:
# !wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract6/pr6_1.txt
````

````python
with open('pr6_1.txt', 'r', encoding='utf-8') as f:
    s = f.read().strip()

count = 0
for i in range(2, len(s)):
    if s[i] == 'D' and s[i-1].isdigit() and s[i-2].isdigit():
        count += 1

print(count)
````

### Вопрос 4
**Что не так с приведенным ниже фрагментом кода?**
```python
animal = 'Тигр'
animal(0) = 'Л'
```

В этом коде две критические ошибки:
1. **Синтаксис индексации:** для обращения по индексу используются квадратные скобки `[]`. Круглые скобки `animal(0)` означают вызов функции/объекта, что вызовет ошибку `TypeError: 'str' object is not callable`.
2. **Неизменяемость строк (immutability):** строки в Python являются неизменяемым типом данных (`immutable`). Даже запись с квадратными скобками `animal[0] = 'Л'` приведёт к ошибке `TypeError: 'str' object does not support item assignment`.

### Вопрос 6
**Что покажет приведенный ниже фрагмент кода?**
```python
mystring = 'abcdefg'
print(mystring[3:])
```

Выведет: **`defg`**  
*Пояснение:* срез `[3:]` берет символы, начиная с индекса 3 (`'d'`) и до конца строки.

### Вопрос 7
**Что покажет приведенный ниже фрагмент кода?**
```python
mystring = 'abcdefg'
print(mystring[:3])
```

Выведет: **`abc`**  
*Пояснение:* срез `[:3]` берет символы с начала строки (индекс 0) до индекса 3 не включая (индексы 0, 1, 2 — символы `'a'`, `'b'`, `'c'`).
