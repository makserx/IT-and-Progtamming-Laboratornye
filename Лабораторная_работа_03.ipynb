# Лабораторная работа 3. Пользовательские функции

## Задание

Напишите функцию для вычисления суммы квадратов натуральных чисел от **a** до **b**, где **a** и **b** – натуральные числа. Продемоснстрируйте использование объявленной функции.

Напишите функцию, которая по трем переданным натуральным
числам возвращает могут ли эти числа являться членами
арифметической прогрессии. Если «да», то функция возвращает
True, в противном случае – False.

## Решение

### Объявление функции

````python
def sum_ab(a, b):
  s = 0
  for i in range(a, b+1):
    s+=i*i
  return s
````

### Демонстрация использования функции

````python
x, y = 1, 2
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 500, 50000
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 268, 3258
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
````

Решите пример в соответствии с индивидуальным заданием

````python
def is_ariphmetic(a,b,c):
	nums = [a,b,c]
	nums1 = sorted(nums)
	return nums1[0] + nums1[2] == 2 * nums1[1]

a,b,c = 1,2,3
print('Могут ли числа являться арифметической прогрессией?', is_ariphmetic(a,b,c))
a,b,c = 100, 50, 365
print('Могут ли числа являться арифметической прогрессией?', is_ariphmetic(a,b,c))
a,b,c = 40,30,20
print('Могут ли числа являться арифметической прогрессией?', is_ariphmetic(a,b,c))
````

1. Чем отличается функция с возвратом значения от функции без возврата значения?
Функция с возвратом значения после выполнения возвращает результат в точку вызова с помощью оператора return (например, return result). Функция без возврата значения (процедура) выполняет действия, но не возвращает результат — фактически она возвращает None.

2. Что такое библиотечная функция?
Это функция, которая уже написана и входит в состав библиотеки (модуля) языка программирования. Её можно подключать и использовать в своей программе, не разрабатывая заново (например, math.sqrt(), random.randint()).

3. Почему библиотечные функции похожи на «чёрные ящики»?
Потому что программисту не нужно знать, как именно функция устроена внутри. Ему достаточно знать её имя, назначение, принимаемые аргументы и возвращаемый результат. Внутренняя реализация скрыта.

4. Что делает инструкция х = random.randint(1, 100)?
Генерирует случайное целое число в диапазоне от 1 до 100 включительно и присваивает его переменной x.

5. Что делает инструкция print(random.randint(1, 20))?
Генерирует случайное целое число в диапазоне от 1 до 20 включительно и выводит его на экран.

6. Для чего используется вызов random.seed(20)?
Для инициализации генератора случайных чисел конкретным начальным значением (зерном). Это делает последовательность «случайных» чисел воспроизводимой — при одном и том же seed генератор будет выдавать одну и ту же последовательность.

7. Каким образом функции помогают повторно использовать код?
Один раз написанную функцию можно вызывать многократно из разных мест программы (и даже из других программ), не переписывая её код заново. Это сокращает объём кода, упрощает его сопровождение и уменьшает количество ошибок.

9. Как у функции задать параметры по умолчанию?
В заголовке функции параметру присваивается значение прямо в списке параметров:
```python

def greet(name='Гость'):
    print('Привет,', name)
```
Если аргумент не передан, используется значение по умолчанию.

10. Что такое рекурсия?
Это приём, при котором функция вызывает саму себя. Рекурсивная функция обязательно должна иметь базовый случай (условие остановки), иначе она будет вызываться бесконечно и приведёт к переполнению стека. Пример — вычисление факториала:
```python

def fact(n):
    if n == 0:
        return 1
    return n * fact(n - 1)
```
