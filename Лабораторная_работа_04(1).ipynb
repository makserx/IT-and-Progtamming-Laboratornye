## Обработка символьной последовательности

Строки, подобно списка, являются коллекциями. Только в качестве элементов последовательности выступают символы. Для строк также используются функции нахождения длины `len()`, операция индексации `[]`

Например, можно посимвольно обходить строку, используя цикл `for` и индекс строки:

````python
# Вывод первых 10 символов строки с одновременным анализом буква это или цифра
for i in range(10):
  c = s[i]
  if c.isdigit():
    t = 'цифра'
  else:
    t='буква'
  print(c, t)
````

## Решение задачи 1

Будем посимвольно обходимть строку и накапливать символы-цифры в строке `target`. Если цифровая последовательность прервется, то конвертируем `target` в число `int` и проверяем является ли число четным и не превысило ли оно максимум

````python
target=''
maxx=0
for c in s:
  if c.isdigit():
    target+=c
  else:
    if target!='':
      x = int(target)
      maxx=max(x, maxx)
      target=''

print('Максимальное четное число = %d' % maxx)
````

## Решение задачи 2

Решение данной задачи осуществляется аналогично предыдущей. Можно даже вывести самуюдлинную буквенную строку

````python
target=''
maxx=0
maxs=''
for c in s:
  if c.isalpha():
    target+=c
  else:
    if len(target)>maxx:
      maxx=len(target)
      maxs=target
      target=''

print('Максимальная длина буквенной строки = %d' % maxx)
print('Сама строка:', maxs)
````

## Решение задачи 3

Для решения данной задачи будем использовать коллекцию словарь `dict()`. Данная коллекция содержит элемкнты в виде пар `ключ : значение`. Доступ к элементам осуществляется по ключу с использованием операции ндексации.

Заполним словарь d, который после прохода по строке должен принять вид:
d = {'A' : 10, 'B' : 127, '3' : 217, ... }
то есть в качестве ключа будем использовать символ, а вкачестве значения - количество таких символов в строке:

````python
# Создаем пустой словарь
d={}
# Проходим по строке и заполняем словарь
for c in s:
  d[c] = d.get(c, 0) + 1

# Выводим получившийся словарь:
print(d)
````

В получившемся словаре необходимо найти самый частотный символ. Для этого не будем обходить словарь, а просто воспользуемся встроенной функцией `max()`, которую применим к коллекции `values()` словаря

````python
max_val = max(d.values())
print(max_val)
````

Выполним обход словаря и напечатаем ответ к данной задаче:

````python
for k in d:
  if d[k]==max_val:
    print(k, end=' ')
print(max_val)
````

Ответы на контрольные вопросы:
1. Назначение, сходства и различия read(), readline(), readlines()
    
read()	Читает весь файл целиком (или n символов, если указать аргумент)
readline()	Читает одну строку (включая \n)
readlines()	Читает все строки файла

Сходства:

    Все три метода читают данные из текстового файла.
    Все они перемещают позицию считывания файла вперёд.
    Все возвращают строковые данные (или список строк).
    После достижения конца файла возвращают пустые значения ('' или []).

Различия:

    read() — читает всё сразу.

    readline() — читает по одной строке.

    readlines() — читает весь файл, но возвращает список строк.

2. Как прочитать многострочный файл без read(), readline(), readlines()

Файловый объект в Python является итерируемым, поэтому можно использовать цикл for:
```python

with open('file.txt', 'r') as f:
    for line in f:
        print(line.rstrip())
```
Также можно использовать list(f) для получения списка строк или f.read().splitlines(), но последнее уже использует read().
3. Задача закрытия файла

Закрытие файла:

    освобождает системные ресурсы (файловый дескриптор);

    гарантирует, что все буферизованные данные записаны на диск;

    предотвращает утечки ресурсов и возможные ошибки при дальнейшей работе;

    снимает блокировку файла (если она была).

В Python рекомендуется использовать конструкцию with open(...), которая автоматически закрывает файл.
4. Что такое позиция считывания файла?

Позиция считывания — это указатель на место в файле, с которого будет прочитан следующий символ или байт. При открытии файла в режиме чтения ('r') позиция изначально находится в начале файла (смещение 0).
5. Режим для записи с сохранением содержимого

Нужен режим дозаписи — 'a' (append). Данные записываются в конец файла, существующее содержимое остаётся нетронутым.
```python

with open('file.txt', 'a') as f:
    f.write('новая строка\n')
```
6. Что такое исключение?

Исключение — это событие, возникающее во время выполнения программы, которое нарушает нормальный ход её выполнения. Оно сигнализирует об ошибке или нештатной ситуации (например, деление на ноль, отсутствие файла, неверный тип данных). Исключения можно перехватывать и обрабатывать с помощью try/except.
7. Что происходит, если исключение не обрабатывается?

Если исключение вызвано, но не обработано инструкцией try/except, программа аварийно завершается, а в консоль выводится трассировка стека (traceback) с информацией об ошибке и месте её возникновения.
8. Исключение при открытии несуществующего файла

При попытке открыть несуществующий файл в режиме чтения возникает исключение:
```text

FileNotFoundError

(подкласс OSError).
```    
9. Исключение при float() для нечисловой строки

При конвертации нечислового строкового значения в число возникает исключение:
```text

ValueError
```
Пример:
```python

float('abc')  # ValueError: could not convert string to float: 'abc'
```
10. Что происходит при открытии несуществующего файла в режиме дозаписи?

Если файл не существует и программа открывает его в режиме дозаписи ('a'), файл будет создан автоматически, и запись начнётся с начала (с нулевой позиции). Исключения не возникает.
```python

with open('new_file.txt', 'a') as f:
    f.write('данные\n')  # файл будет создан
```
